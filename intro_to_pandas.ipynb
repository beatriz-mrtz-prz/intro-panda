## 2. Series y Dataframes

Pandas tiene dos estructuras de datos principales: Series y DataFrames.

- **Series**: es un array unidimensional. Es similar a una columna en una tabla de Excel.
- **DataFrame**: es una estructura de datos bidimensional con filas y columnas con etiquetas. Es similar a una tabla de Excel.

### 2.1 Series

Para crear una Serie, puedes pasar una lista de elementos y Pandas creará una Serie con índices numéricos.

````python
serie = pd.Series([1, 2, 3, 4, 5])
serie
````

### 2.2 DataFrames

Para crear un DataFrame, puedes pasar un diccionario de listas, donde cada clave es el nombre de la columna y cada valor es una lista de elementos.

````python
data = {
    'name': ['María', 'Daniel', 'Marisa', 'David', 'Sabrina'],
    'age': [16, 7, 29, None, 33], 
    'city': ['Madrid', 'Madrid', 'Málaga', 'Málaga', 'Valladolid'],
    'height': [None, 1.20, 1.60, 2.01, 1.83]
}

df = pd.DataFrame(data)
df
````

Obtén los datos de la columna 'name' ¿Cual es el tipo de dato de esta columna?

🌶️ Recuera que una Serie se puede tratar como un array de numpy, por tanto, obtén los datos de la columna "age" y calcula la media. ¿Coincide con el resultado que arroja pandas en la celda superior? (💡 en `numpy` la media se calcula usando `np.mean`)

Una Serie se comporta como un array de NumPy, así que `np.mean()` funciona directamente sobre ella. Ambos métodos **ignoran los valores nulos**, por eso el resultado coincide con el de `describe()`.

````python
import numpy as np

age_column = people_df["age"]

mean_age_with_numpy = np.mean(age_column)
mean_age_with_pandas = age_column.mean()

print(f"Media de edad con numpy:  {mean_age_with_numpy:.2f} años")
print(f"Media de edad con pandas: {mean_age_with_pandas:.2f} años")
print(f"¿Coinciden ambos resultados? "
      f"{np.isclose(mean_age_with_numpy, mean_age_with_pandas)} "
      f"(los dos descartan los valores nulos)")
````

Ahora calcula la edad máxima y mínima de la columna "age".

Los métodos `max()` y `min()` de una Serie devuelven el valor extremo, también ignorando los nulos.

````python
max_age = people_df["age"].max()
min_age = people_df["age"].min()

print(f"La edad máxima es de {max_age:.0f} años "
      f"y la edad mínima es de {min_age:.0f} años.")
````

🌶️ Una vez obtenida la edad mínima, ¿Cual es el nombre de dicha persona?

`idxmin()` no devuelve el valor mínimo, sino **la etiqueta de la fila** donde se encuentra. Con esa etiqueta podemos recuperar cualquier otro dato de la fila usando `loc`.

````python
min_age_row_label = people_df["age"].idxmin()
youngest_person_name = people_df.loc[min_age_row_label, "name"]

print(f"La persona más joven es {youngest_person_name}, "
      f"con {min_age:.0f} años (fila {min_age_row_label}).")
````

Filtra el DataFrame: muestra las filas donde la edad sea mayor a 25.

Creamos primero una máscara booleana con la condición y después la usamos para filtrar. Separar la condición en su propia variable hace el código más legible y nos permite reutilizarla más adelante.

````python
is_older_than_25 = people_df["age"] > 25
older_than_25_df = people_df[is_older_than_25]

print(f"Personas mayores de 25 años: {len(older_than_25_df)}")
older_than_25_df
````

Filtra el DataFrame: muestra las filas donde la gente sea de 'Málaga'. (💡 utiliza el operador `==`)

Misma lógica, pero comparando texto con el operador `==`.

````python
is_from_malaga = people_df["city"] == "Málaga"
malaga_df = people_df[is_from_malaga]

print(f"Personas de Málaga: {len(malaga_df)}")
malaga_df
````

🌶️ Filtra el Dataframe: filas donde la gente sea de 'Málaga' y mayor de 25. (💡 para sumar dos condiciones se utiliza el operador `and` que se representa en python con `&`, i.e.: 

```python
compose_condition = condition1 & condition2
```

Combinamos las dos máscaras anteriores con `&` (AND elemento a elemento). Importante: los registros con edad nula quedan fuera, porque cualquier comparación con `NaN` devuelve `False`.

````python
is_from_malaga_and_older_than_25 = is_from_malaga & is_older_than_25
malaga_older_than_25_df = people_df[is_from_malaga_and_older_than_25]

print(f"Personas de Málaga mayores de 25 años: {len(malaga_older_than_25_df)}")
malaga_older_than_25_df
````

Obtén la máscara de los valores que son nulos en el DataFrame.

`isnull()` devuelve un DataFrame del mismo tamaño con `True` donde hay un nulo. Como `True` vale 1, una doble suma nos da el total de nulos.

````python
null_values_mask = people_df.isnull()
total_null_values = int(null_values_mask.sum().sum())

print(f"El DataFrame contiene {total_null_values} valores nulos en total.")
null_values_mask
````

Ahora obtén la máscara de los valores que NO son nulos en el DataFrame. (💡 busca en la documentación de pandas, o en google, cómo hacerlo)

`notnull()` (o `notna()`) es la máscara inversa: `True` donde **sí** hay dato.

````python
not_null_values_mask = people_df.notnull()
total_not_null_values = int(not_null_values_mask.sum().sum())

print(f"El DataFrame contiene {total_not_null_values} valores no nulos.")
not_null_values_mask
````

Elimina los valores nulos del DataFrame.

`dropna()` elimina las filas que tengan **algún** valor nulo. No modifica el DataFrame original: devuelve uno nuevo, por eso lo guardamos en otra variable.

````python
people_without_nulls_df = people_df.dropna()
removed_rows = len(people_df) - len(people_without_nulls_df)

print(f"Se han eliminado {removed_rows} filas con valores nulos: "
      f"pasamos de {len(people_df)} a {len(people_without_nulls_df)} registros.")
people_without_nulls_df
````

Rellena todos los valores nulos con 1.

`fillna(valor)` sustituye todos los nulos del DataFrame por el valor indicado.

````python
people_filled_with_ones_df = people_df.fillna(1)

print("Todos los valores nulos han sido sustituidos por 1:")
people_filled_with_ones_df
````

Ahora rellena sólo los valores nulos de la columna 'age' con 10.

Si solo queremos tratar una columna, aplicamos `fillna()` sobre esa Serie y reasignamos el resultado. Trabajamos sobre una copia para no alterar el DataFrame original.

````python
people_filled_age_df = people_df.copy()
people_filled_age_df["age"] = people_filled_age_df["age"].fillna(10)

print("Los valores nulos de la columna 'age' se han rellenado con 10:")
people_filled_age_df
````

🌶️ Ahora rellena los valores nulos de la columna 'height' con la media de dicha columna

Rellenar con la media es una estrategia de imputación muy habitual: mantiene la media de la columna intacta y evita perder filas. Calculamos la media (que ignora los nulos) y la usamos como valor de relleno.

````python
mean_height = people_df["height"].mean()

people_filled_height_df = people_df.copy()
people_filled_height_df["height"] = people_filled_height_df["height"].fillna(mean_height)

print(f"La altura media es de {mean_height:.2f} m. "
      f"Los valores nulos de 'height' se han rellenado con ese valor:")
people_filled_height_df
````

### B.1 ¿Qué porcentaje de pasajeros sobrevivió?

La columna `Survived` está codificada como 0 (no sobrevivió) y 1 (sobrevivió). Al ser una variable binaria, **su media es directamente la tasa de supervivencia**: basta con multiplicarla por 100.

````python
total_passengers = len(titanic_df)
survivors_count = int(titanic_df["Survived"].sum())
survival_rate = titanic_df["Survived"].mean() * 100

print(f"De los {total_passengers} pasajeros del dataset, sobrevivieron {survivors_count}.")
print(f"La tasa de supervivencia es del {survival_rate:.2f}%.")
````

### B.2 ¿Cuál es la edad media de los pasajeros?

Calculamos la media de la columna `Age`. Recordemos que `mean()` ignora los nulos, así que la media se calcula solo sobre los pasajeros con edad conocida: lo indicamos explícitamente en el resultado para no dar una cifra engañosa.

````python
age_column = titanic_df["Age"]
mean_age = age_column.mean()
passengers_with_known_age = int(age_column.notnull().sum())

print(f"La edad media de los pasajeros es de {mean_age:.2f} años, "
      f"calculada sobre los {passengers_with_known_age} pasajeros con edad conocida "
      f"(de un total de {total_passengers}).")
````

### B.3 ¿Cuántos pasajeros había en cada clase?

`value_counts()` cuenta cuántas veces se repite cada valor de la columna `Pclass`. Ordenamos por clase con `sort_index()` para leerlo de 1ª a 3ª.

````python
passengers_per_class = titanic_df["Pclass"].value_counts().sort_index()

print("Número de pasajeros por clase:")
for passenger_class, passengers_count in passengers_per_class.items():
    class_percentage = passengers_count / total_passengers * 100
    print(f"  - {passenger_class}ª clase: {passengers_count} pasajeros "
          f"({class_percentage:.1f}% del total)")
````

### B.4 Edad media de quienes sobrevivieron y de quienes no

Aquí usamos `groupby()`: agrupamos las filas por el valor de `Survived` y aplicamos la media a la columna `Age` dentro de cada grupo.

````python
mean_age_by_survival = titanic_df.groupby("Survived")["Age"].mean()

mean_age_survivors = mean_age_by_survival[1]
mean_age_non_survivors = mean_age_by_survival[0]

print(f"La edad media de los pasajeros que SÍ sobrevivieron es de "
      f"{mean_age_survivors:.2f} años.")
print(f"La edad media de los que NO sobrevivieron es de "
      f"{mean_age_non_survivors:.2f} años.")
print(f"Diferencia: {abs(mean_age_survivors - mean_age_non_survivors):.2f} años.")
````

### B.5 Edad máxima de quienes sobrevivieron y de quienes no

Mismo `groupby`, cambiando la función agregada por `max()`.

````python
max_age_by_survival = titanic_df.groupby("Survived")["Age"].max()

max_age_survivors = max_age_by_survival[1]
max_age_non_survivors = max_age_by_survival[0]

print(f"El pasajero de más edad que sobrevivió tenía {max_age_survivors:.0f} años.")
print(f"El pasajero de más edad que no sobrevivió tenía {max_age_non_survivors:.0f} años.")
````

### B.6 Extra: ¿influyeron el sexo y la clase en la supervivencia?

Aplicando la misma idea del apartado B.1 (la media de una columna binaria es una tasa) sobre grupos, podemos comparar la supervivencia por sexo y por clase.

````python
survival_rate_by_sex = titanic_df.groupby("Sex")["Survived"].mean() * 100

print("Tasa de supervivencia por sexo:")
for sex, rate in survival_rate_by_sex.items():
    print(f"  - {sex}: {rate:.2f}%")
````

````python
survival_rate_by_class = titanic_df.groupby("Pclass")["Survived"].mean() * 100

print("Tasa de supervivencia por clase:")
for passenger_class, rate in survival_rate_by_class.items():
    print(f"  - {passenger_class}ª clase: {rate:.2f}%")
````

````python
survival_rate_by_class_and_sex = (
    titanic_df.groupby(["Pclass", "Sex"])["Survived"].mean() * 100
).round(2)

print("Tasa de supervivencia (%) cruzando clase y sexo:")
survival_rate_by_class_and_sex.unstack()
````
